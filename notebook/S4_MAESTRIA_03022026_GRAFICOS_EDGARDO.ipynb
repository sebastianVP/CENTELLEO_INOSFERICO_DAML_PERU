### **1.3 PRE-PROCESAMIENTO**

#### **1.3.1 FILTRO DE ANGULO DE ELEVACION-UTILIZAMOS UMBRAL DE 30**

1. Minimizar el efecto multipath:

    * Cuando un satélite está cerca del horizonte (baja elevación), la señal GNSS recorre más atmósfera y tiene más probabilidad de reflejarse en superficies terrestres (edificios, suelo, agua), introduciendo interferencias multipath.

    * Estas reflexiones distorsionan las medidas de fase y amplitud.

2. Reducción de la distorsión atmosférica:

    * A bajas elevaciones, la señal atraviesa más atmósfera, lo que introduce mayor error por refracción, dispersión e interacciones con la ionósfera y la troposfera.

3. Estándares y recomendaciones en investigación:
    * Muchos estudios académicos y operacionales usan 30° como valor estándar.


**NOTA:** El umbral de 30° es un compromiso práctico: filtra señales de mala calidad (por baja elevación) sin descartar demasiada información útil. En contextos más exigentes, algunos usan 40°, y en otros más permisivos, 20°.

**FUENTES**
1. ITU-R P.681-11 (International Telecommunication Union):
      * Define la recomendación de evitar ángulos bajos de elevación para comunicaciones satelitales por sus altos niveles de interferencia.

2. Investigaciones de scintilación ionosférica:
      * En artículos como:
    “Ionospheric Scintillation Effects on GNSS Signals in Equatorial Regions”
    (Kintner, Ledvina & de Paula, 2007)
      * Se recomienda trabajar con elevaciones > 30° para evitar artefactos en la caracterización de cintilaciones.

```python
# FILTRO DE 30 ° grados
#s4_param = s4_param[s4_param["Elevacion"]>30]
s4_param = s4_param[s4_param["ELEVATION"]>30]

print("Cantidad del dataset:" , len(s4_param))
s4_param.head()
```

#### **1.3.2 DETECCIÓN Y ELIMINACION DE OUTLIERS**

El criterio empleado esta basado en la continuidad del parametro.

1. Ordenaremos por satélite y tiempo el dataset
2. Umbral de diferencia relativa(ajustable).
3. Crearemos la función **detectar_outliers** para detectar outliers sin eliminarlos.

image.png

```python
import numpy as np
# UMBRAL DE DIFERENCIA RELATIVA(AJUSTABLE)
threshold_ratio =2

# DATAFRAME PARA ALMACENAR OUTLIERS
outliers_detectados = pd.DataFrame()

# FUNCION PARA DETECTAR OUTLIERS SIN ELIMINARLOS AUN
def detectar_outliers(grupo):
  s4            = grupo['S4'].values
  #tiempos       = grupo['Tiempo'].values
  tiempos       = grupo['TIME'].values

  outlier_flags = [False]*len(s4)

  for i in range(1,len(s4)-1):
    prev             = s4[i-1]
    curr             = s4[i]
    nxt              = s4[i+1]
    promedio_vecinos =  (prev+nxt)/2.0

    if promedio_vecinos >0 and abs(curr-promedio_vecinos>threshold_ratio*promedio_vecinos):
      outlier_flags[i] = True

  grupo['is_Outlier'] = outlier_flags
  return grupo
```

```python
# Ordenar por satélite y tiempo
#s4_param.sort_values(by=['ID_Satelite','Tiempo'],inplace=True)
s4_param.sort_values(by=['ID_SATELLITE','TIME'],inplace=True)

s4_param.head()
```

# **1RA VERSION**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Convert UTC → Local Time (UTC - 5)
# ============================================================
def to_local_time(df, time_col):
    df = df.copy()
    df[time_col] = pd.to_datetime(df[time_col]) - pd.Timedelta(hours=5)
    return df


# ============================================================
# Continuous nocturnal minute in LOCAL TIME
# Night associated to DAY D:
#   D 18:00  →  23:59
# D+1 00:00  →  06:00
# ============================================================
def nocturnal_minute_lt(dt):
    h = dt.hour
    m = dt.minute

    if h >= 18:
        return (h - 18) * 60 + m
    elif h <= 6:
        return (h + 6) * 60 + m
    else:
        return np.nan


# ============================================================
# Assign nocturnal DAY (00–06 → previous day)
# ============================================================
def nocturnal_day(dt):
    if dt.hour <= 6:
        return (dt - pd.Timedelta(days=1)).dayofyear
    else:
        return dt.dayofyear


# ============================================================
# Annual S4 plot (Day vs nocturnal minute)
# ============================================================
def plot_s4_year_minute(df, year):

    # --------------------------------------------------------
    # 1. Convert to Local Time
    # --------------------------------------------------------
    df = to_local_time(df, "Tiempo")

    # --------------------------------------------------------
    # 2. FILTER BY YEAR (LOCAL TIME — FIRST RULE)
    # --------------------------------------------------------
    df_year = df[df["Tiempo"].dt.year == year].copy()

    if df_year.empty:
        raise ValueError(f"No data available for year {year}")

    # --------------------------------------------------------
    # 3. Nocturnal variables (inside the year)
    # --------------------------------------------------------
    df_year["Nocturnal_Minute"] = df_year["Tiempo"].apply(nocturnal_minute_lt)
    df_year["Nocturnal_Day"] = df_year["Tiempo"].apply(nocturnal_day)

    df_year = df_year.dropna(subset=["Nocturnal_Minute"])

    if df_year.empty:
        raise ValueError(f"No nocturnal data available for year {year}")

    # --------------------------------------------------------
    # 4. FIRST and LAST nocturnal day WITH DATA (KEY POINT)
    # --------------------------------------------------------
    first_day = int(df_year["Nocturnal_Day"].min())
    last_day  = int(df_year["Nocturnal_Day"].max())

    first_date = pd.Timestamp(year=year, month=1, day=1) + pd.Timedelta(days=first_day - 1)
    last_date  = pd.Timestamp(year=year, month=1, day=1) + pd.Timedelta(days=last_day - 1)

    # --------------------------------------------------------
    # 5. Axis definitions (RECORTADAS AL RANGO REAL)
    # --------------------------------------------------------
    nocturnal_minutes = np.arange(0, 721)          # 18:00 → 06:00
    days_of_year = np.arange(first_day, last_day + 1)

    # --------------------------------------------------------
    # 6. Matrix: Minute × Day (recortada)
    # --------------------------------------------------------
    Z = np.full((len(nocturnal_minutes), len(days_of_year)), np.nan)

    for _, row in df_year.iterrows():
        i = int(row["Nocturnal_Minute"])
        j = int(row["Nocturnal_Day"] - first_day)
        if 0 <= i < Z.shape[0] and 0 <= j < Z.shape[1]:
            Z[i, j] = row["S4"]

    # --------------------------------------------------------
    # 7. Month ticks (FROM FIRST DATA → LAST DATA ONLY)
    # --------------------------------------------------------
    dates = pd.date_range(start=first_date, end=last_date, freq="D")

    month_positions = []
    month_labels = []

    for m in sorted(dates.month.unique()):
        idx = np.where(dates.month == m)[0]
        month_positions.append(idx[0])
        month_labels.append(dates[idx[0]].strftime("%b"))

    # --------------------------------------------------------
    # 8. Plot
    # --------------------------------------------------------
    fig, ax = plt.subplots(figsize=(16, 6))

    im = ax.imshow(
        Z,
        aspect="auto",
        origin="lower",
        cmap="jet",
        vmin=0.0,
        vmax=0.6
    )

    # Y-axis: Local Time
    hour_ticks = np.arange(0, 721, 60)
    hour_labels = ["18","19","20","21","22","23",
                   "00","01","02","03","04","05","06"]

    ax.set_yticks(hour_ticks)
    ax.set_yticklabels(hour_labels)
    ax.set_ylabel("Local Time (LT)")

    # X-axis: Months (NO extra months)
    ax.set_xticks(month_positions)
    ax.set_xticklabels(month_labels)
    ax.set_xlabel("Month")

    ax.set_title(
        f"Nocturnal Ionospheric Scintillation Index (S4)\n"
        f"Daily–Minute Resolution (18:00–06:00 LT) – {year}",
        fontsize=12
    )

    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label("S4 Index")

    plt.tight_layout()
    plt.show()


# ============================================================
# USAGE
# ============================================================
s4_param_JIC["Tiempo"] = pd.to_datetime(s4_param_JIC["Tiempo"])
plot_s4_year_minute(s4_param_JIC, 2025)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Convert UTC → Local Time (UTC - 5)
# ============================================================
def to_local_time(df, time_col):
    df = df.copy()
    df[time_col] = pd.to_datetime(df[time_col]) - pd.Timedelta(hours=5)
    return df


# ============================================================
# Continuous nocturnal minute in LOCAL TIME
# Night associated to DAY D:
#   D 18:00  →  23:59
# D+1 00:00  →  06:00
# ============================================================
def nocturnal_minute_lt(dt):
    h = dt.hour
    m = dt.minute

    if h >= 18:
        return (h - 18) * 60 + m
    elif h <= 6:
        return (h + 6) * 60 + m
    else:
        return np.nan


# ============================================================
# Assign nocturnal DAY (00–06 → previous day)
# ============================================================
def nocturnal_day(dt):
    if dt.hour <= 6:
        return (dt - pd.Timedelta(days=1)).dayofyear
    else:
        return dt.dayofyear


# ============================================================
# Annual S4 plot (Day vs nocturnal minute)
# ============================================================
def plot_s4_year_minute(df, year):

    # --------------------------------------------------------
    # 1. Convert to Local Time
    # --------------------------------------------------------
    df = to_local_time(df, "Tiempo")

    # --------------------------------------------------------
    # 2. FILTER BY YEAR (FIRST RULE)
    # --------------------------------------------------------
    df_year = df[df["Tiempo"].dt.year == year].copy()

    if df_year.empty:
        raise ValueError(f"No data available for year {year}")

    # --------------------------------------------------------
    # 3. Nocturnal variables
    # --------------------------------------------------------
    df_year["Nocturnal_Minute"] = df_year["Tiempo"].apply(nocturnal_minute_lt)
    df_year["Nocturnal_Day"] = df_year["Tiempo"].apply(nocturnal_day)
    df_year = df_year.dropna(subset=["Nocturnal_Minute"])

    if df_year.empty:
        raise ValueError(f"No nocturnal data available for year {year}")

    # --------------------------------------------------------
    # 4. FIRST & LAST REAL DATA (MONTH-BASED LOGIC)
    # --------------------------------------------------------
    first_datetime = df_year["Tiempo"].min()
    last_datetime  = df_year["Tiempo"].max()

    first_month_date = first_datetime.replace(day=1)
    last_month_date  = last_datetime.replace(day=1) + pd.offsets.MonthEnd(0)

    # Full daily range INCLUDING missing days (NaN filling)
    full_dates = pd.date_range(
        start=first_month_date,
        end=last_month_date,
        freq="D"
    )

    # Map dates to column indices
    date_to_idx = {d.date(): i for i, d in enumerate(full_dates)}

    nocturnal_minutes = np.arange(0, 721)  # 18:00 → 06:00
    Z = np.full((len(nocturnal_minutes), len(full_dates)), np.nan)

    # --------------------------------------------------------
    # 5. Fill matrix (missing days stay NaN)
    # --------------------------------------------------------
    for _, row in df_year.iterrows():
        night_date = row["Tiempo"].date()
        if row["Tiempo"].hour <= 6:
            night_date = (row["Tiempo"] - pd.Timedelta(days=1)).date()

        i = int(row["Nocturnal_Minute"])
        j = date_to_idx.get(night_date, None)

        if j is not None and 0 <= i < Z.shape[0]:
            Z[i, j] = row["S4"]

    # --------------------------------------------------------
    # 6. Month ticks (ONLY months with real data)
    # --------------------------------------------------------
    months = pd.date_range(
        start=first_month_date,
        end=last_month_date,
        freq="MS"
    )

    month_positions = [
        (m.date() - full_dates[0].date()).days
        for m in months
    ]

    month_labels = [m.strftime("%b") for m in months]

    # --------------------------------------------------------
    # 7. Plot
    # --------------------------------------------------------
    fig, ax = plt.subplots(figsize=(16, 6))

    im = ax.imshow(
        Z,
        aspect="auto",
        origin="lower",
        cmap="jet",
        vmin=0.0,
        vmax=0.6
    )

    # Y-axis: Local Time
    hour_ticks = np.arange(0, 721, 60)
    hour_labels = ["18","19","20","21","22","23",
                   "00","01","02","03","04","05","06"]

    ax.set_yticks(hour_ticks)
    ax.set_yticklabels(hour_labels)
    ax.set_ylabel("Local Time (LT)")

    # X-axis: Months (FULL months, no extra)
    ax.set_xticks(month_positions)
    ax.set_xticklabels(month_labels)
    ax.set_xlabel("Month")

    ax.set_title(
        f"Nocturnal Ionospheric Scintillation Index (S4)\n"
        f"Daily–Minute Resolution (18:00–06:00 LT) – {year}",
        fontsize=12
    )

    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label("S4 Index")

    plt.tight_layout()
    plt.show()


# ============================================================
# USAGE
# ============================================================
s4_param_JIC["Tiempo"] = pd.to_datetime(s4_param_JIC["Tiempo"])
plot_s4_year_minute(s4_param_JIC, 2025)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Convert UTC → Local Time (UTC - 5)
# ============================================================
def to_local_time(df, time_col):
    df = df.copy()
    df[time_col] = pd.to_datetime(df[time_col]) - pd.Timedelta(hours=5)
    return df


# ============================================================
# Continuous nocturnal minute in LOCAL TIME
# ============================================================
def nocturnal_minute_lt(dt):
    h, m = dt.hour, dt.minute

    if h >= 18:
        return (h - 18) * 60 + m
    elif h < 6:                      # ⬅️ CAMBIO CLAVE
        return (h + 6) * 60 + m
    elif h == 6 and m == 0:           # ⬅️ SOLO 06:00
        return 720
    else:
        return np.nan


# ============================================================
# Assign nocturnal DAY
# ============================================================
def nocturnal_day(dt):
    if dt.hour <= 6:
        return (dt - pd.Timedelta(days=1)).dayofyear
    return dt.dayofyear


# ============================================================
# Annual S4 plot
# ============================================================
def plot_s4_year_minute(df, year, station_name="Station", save_fig=False, out_dir="./"):

    df = to_local_time(df, "Tiempo")
    df_year = df[df["Tiempo"].dt.year == year].copy()

    if df_year.empty:
        raise ValueError(f"No data available for year {year}")

    df_year["Nocturnal_Minute"] = df_year["Tiempo"].apply(nocturnal_minute_lt)
    df_year = df_year.dropna(subset=["Nocturnal_Minute"])

    if df_year.empty:
        raise ValueError("No nocturnal data available")

    # ---- Date limits
    first_dt = df_year["Tiempo"].min()
    last_dt  = df_year["Tiempo"].max()

    first_month = first_dt.replace(day=1)
    last_month  = last_dt.replace(day=1) + pd.offsets.MonthEnd(0)

    full_dates = pd.date_range(first_month, last_month, freq="D")
    date_to_idx = {d.date(): i for i, d in enumerate(full_dates)}

    nocturnal_minutes = np.arange(0, 721)
    Z = np.full((len(nocturnal_minutes), len(full_dates)), np.nan)

    # ---- Fill matrix
    for _, row in df_year.iterrows():
        night_date = row["Tiempo"].date()
        if row["Tiempo"].hour <= 6:
            night_date = (row["Tiempo"] - pd.Timedelta(days=1)).date()

        i = int(row["Nocturnal_Minute"])
        j = date_to_idx.get(night_date)

        if j is not None and 0 <= i < Z.shape[0]:
            Z[i, j] = row["S4"]

    # ---- Month ticks
    months = pd.date_range(first_month, last_month, freq="MS")
    month_positions = [(m.date() - full_dates[0].date()).days for m in months]
    month_labels = [m.strftime("%b") for m in months]

    # ========================================================
    # PLOT (ANTI-BLUR CONFIG)
    # ========================================================
    fig, ax = plt.subplots(figsize=(16, 6), dpi=150)

    im = ax.imshow(
        Z,
        origin="lower",
        aspect="auto",
        cmap="jet",
        vmin=0.0,
        vmax=0.6,
        interpolation="nearest",   # 🔥 CLAVE
        extent=[0, Z.shape[1], 0, Z.shape[0]]
    )

    # ---- Y axis
    hour_ticks = np.arange(0, 721, 60)
    hour_labels = ["18","19","20","21","22","23",
                   "00","01","02","03","04","05","06"]
    ax.set_yticks(hour_ticks)
    ax.set_yticklabels(hour_labels)
    ax.set_ylabel("Local Time (LT)")

    # ---- X axis
    ax.set_xticks(month_positions)
    ax.set_xticklabels(month_labels)
    ax.set_xlabel("Month")

    ax.set_title(
        f"Nocturnal Ionospheric Scintillation Index (S4)\n"
        f"Daily–Minute Resolution (18:00–06:00 LT) – {year}\n"
        f"Station: {station_name}",
        fontsize=12
    )

    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label("S4 Index")

    plt.tight_layout()
    os.makedirs(out_dir, exist_ok=True)
    if save_fig:
        station_clean = station_name.replace(" ", "_")
        filename = f"{station_clean}_S4_Nocturnal_{year}.png"
    
        plt.savefig(
            f"{out_dir}/{filename}",
            dpi=300,
            bbox_inches="tight"
        )






    plt.show()


# ============================================================
# USAGE
# ============================================================
s4_param_JIC["Tiempo"] = pd.to_datetime(s4_param_JIC["Tiempo"])
plot_s4_year_minute(s4_param_JIC, 2025, station_name="JICAMARCA",save_fig=True,out_dir="figures")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Convert UTC → Local Time (UTC - 5)
# ============================================================
def to_local_time(df, time_col):
    df = df.copy()
    df[time_col] = pd.to_datetime(df[time_col]) - pd.Timedelta(hours=5)
    return df


# ============================================================
# Continuous nocturnal minute in LOCAL TIME
# ============================================================
def nocturnal_minute_lt(dt):
    h, m = dt.hour, dt.minute

    if h >= 18:
        return (h - 18) * 60 + m
    elif h < 6:
        return (h + 6) * 60 + m
    elif h == 6 and m == 0:
        return 720
    else:
        return np.nan


# ============================================================
# Annual S4 plot
# ============================================================
def plot_s4_year_minute(df, year, station_name="STATION", save_fig=False, out_dir="./"):

    df = to_local_time(df, "TIME")
    df_year = df[df["TIME"].dt.year == year].copy()

    if df_year.empty:
        raise ValueError(f"No data available for year {year}")

    df_year["Nocturnal_Minute"] = df_year["TIME"].apply(nocturnal_minute_lt)
    df_year = df_year.dropna(subset=["Nocturnal_Minute"])

    if df_year.empty:
        raise ValueError("No nocturnal data available")

    # ---- Date limits
    first_dt = df_year["TIME"].min()
    last_dt  = df_year["TIME"].max()

    first_month = first_dt.replace(day=1)
    last_month  = last_dt.replace(day=1) + pd.offsets.MonthEnd(0)

    full_dates = pd.date_range(first_month, last_month, freq="D")
    date_to_idx = {d.date(): i for i, d in enumerate(full_dates)}

    nocturnal_minutes = np.arange(0, 721)
    Z = np.full((len(nocturnal_minutes), len(full_dates)), np.nan)

    # ---- Fill matrix
    for _, row in df_year.iterrows():
        night_date = row["TIME"].date()
        if row["TIME"].hour <= 6:
            night_date = (row["TIME"] - pd.Timedelta(days=1)).date()

        i = int(row["Nocturnal_Minute"])
        j = date_to_idx.get(night_date)

        if j is not None and 0 <= i < Z.shape[0]:
            Z[i, j] = row["S4"]

    # ========================================================
    # MONTH LABELS (SPANISH - UPPERCASE)
    # ========================================================
    months = pd.date_range(first_month, last_month, freq="MS")

    month_positions = [(m.date() - full_dates[0].date()).days for m in months]

    month_labels = [
        m.strftime("%b").upper()
        .replace("JAN", "ENE")
        .replace("FEB", "FEB")
        .replace("MAR", "MAR")
        .replace("APR", "ABR")
        .replace("MAY", "MAY")
        .replace("JUN", "JUN")
        .replace("JUL", "JUL")
        .replace("AUG", "AGO")
        .replace("SEP", "SEP")
        .replace("OCT", "OCT")
        .replace("NOV", "NOV")
        .replace("DEC", "DIC")
        for m in months
    ]

    # ========================================================
    # PLOT
    # ========================================================
    fig, ax = plt.subplots(figsize=(16, 6), dpi=150)

    im = ax.imshow(
        Z,
        origin="lower",
        aspect="auto",
        cmap="jet",
        vmin=0.0,
        vmax=0.6,
        interpolation="nearest",
        extent=[0, Z.shape[1], 0, Z.shape[0]]
    )

    # ---- Y AXIS (UPPERCASE)
    hour_ticks = np.arange(0, 721, 60)
    hour_labels = ["18","19","20","21","22","23",
                   "00","01","02","03","04","05","06"]

    ax.set_yticks(hour_ticks)
    ax.set_yticklabels(hour_labels)

    ax.set_ylabel("LOCAL TIME (LT)", fontsize=12)

    # ---- X AXIS (UPPERCASE)
    ax.set_xticks(month_positions)
    ax.set_xticklabels(month_labels)

    ax.set_xlabel("MONTHS", fontsize=12)

    # ---- TITLE (UPPERCASE)
    ax.set_title(
        f"NOCTURNAL IONOSPHERIC SCINTILLATION INDEX (S4)\n"
        f"DAILY–MINUTE RESOLUTION (18:00–06:00 LT) – {year}\n"
        f"STATION: {station_name}",
        fontsize=12
    )

    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label("S4 INDEX")

    plt.tight_layout()
    os.makedirs(out_dir, exist_ok=True)

    if save_fig:
        station_clean = station_name.replace(" ", "_")
        filename = f"{station_clean}_S4_NOCTURNAL_{year}.png"

        plt.savefig(
            f"{out_dir}/{filename}",
            dpi=300,
            bbox_inches="tight"
        )

    plt.show()
```

#### **1.7 DETECCION DE PICOS Y EVENTOS ANOMALOS**

Detectaremos el **inicio y fin del evento principal**(aumento súbito de S4) que contiene el pico máximo en una señal de centelleo en el rango de 0 a 24 horas.

##### **1.7.1 METODO PROPUESTO**

1. Encontraremos el pico maximo(maximo valor de S4)
2. Definiremos umbral dinamico relativo **(ej. 30% del máximo)**
3. Retrocederemos desde el pico hasta que S4 baje del umbral: MARCAREMOS COMO INICIO.
4. Avanzar dede el pico hasta que S4 vuelva a bajar del umbral: MARCAREMOS COMO FIN.

#### **1.7.2 GRAFICAMENTE APLICACION DE LA DETECCION, INICIO Y FIN**

```python
import pandas as pd
import plotly.graph_objects as go

def detectarEventosPlotly(df_input, inicio, fin, titulo="Detección del evento de centelleo"):
    df_input['Tiempo'] = pd.to_datetime(df_input['Tiempo'])
    df = df_input[(df_input['Tiempo'] >= inicio) & (df_input['Tiempo'] <= fin)].reset_index(drop=True)

    if df.empty:
        print("No hay datos en el rango seleccionado.")
        return

    idx_max = df['S4'].idxmax()
    pico_max = df['S4'].iloc[idx_max]
    umbral = 0.3 * pico_max

    inicio_idx = idx_max
    while inicio_idx > 0 and df['S4'].iloc[inicio_idx] > umbral:
        inicio_idx -= 1

    fin_idx = idx_max
    while fin_idx < len(df) - 1 and df['S4'].iloc[fin_idx] > umbral:
        fin_idx += 1

    inicio_tiempo = df['Tiempo'].iloc[inicio_idx]
    fin_tiempo = df['Tiempo'].iloc[fin_idx]

    print(f"Inicio del evento: {inicio_tiempo}")
    print(f"Fin del evento: {fin_tiempo}")
    print(f"Pico Máximo: {pico_max}")

    fig = go.Figure()

    # Traza principal
    fig.add_trace(go.Scatter(
        x=df['Tiempo'],
        y=df['S4'],
        mode='lines+markers',
        name='S4',
        line=dict(color='blue')
    ))

    # Líneas verticales con add_shape
    for x, color, name in [
        (df['Tiempo'].iloc[idx_max], 'red', 'Pico Máximo'),
        (inicio_tiempo, 'green', 'Inicio'),
        (fin_tiempo, 'orange', 'Fin')
    ]:
        fig.add_shape(
            type="line",
            x0=x, x1=x,
            y0=0, y1=1,  # relativo al eje y
            xref='x',
            yref='paper',  # 'paper' usa 0-1 relativo al alto del gráfico
            line=dict(color=color, width=2, dash='dash')
        )
        # Texto flotante (opcional)
        fig.add_annotation(
            x=x,
            y=1,
            yref="paper",
            text=name,
            showarrow=False,
            font=dict(size=10, color=color),
            align="center",
            bgcolor="rgba(255,255,255,0.6)"
        )

    fig.update_layout(
        title=titulo,
        xaxis_title="Tiempo",
        yaxis_title="S4",
        template="plotly_white",
        height=450,
        showlegend=True
    )

    fig.show()
```

#### **1.7.3 DATAFRAME DE APLICACION DE LA DETECCION, INICIO Y FIN**

Metodo para detectar la fecha, inicio_evento,fin_evento y pico_max.

```python
def detectar_eventos_por_dia(df, umbral_relativo=0.3, mostrar_grafico=False):
    """
    Detecta el evento principal (inicio, fin, pico) para cada fecha única en el DataFrame.

    Parámetros:
        df (pd.DataFrame): Contiene columnas 'Tiempo' y 'S4'.
        umbral_relativo (float): Fracción del pico máximo para determinar el evento.
        mostrar_grafico (bool): Si True, muestra un gráfico por cada fecha.

    Retorna:
        lista de diccionarios con fecha, inicio_evento, fin_evento y pico_max.
    """
    df['Tiempo'] = pd.to_datetime(df['Tiempo'])
    df['Fecha'] = df['Tiempo'].dt.date
    fechas_unicas = sorted(df['Fecha'].unique())

    resultados = []

    for fecha in fechas_unicas:
        df_dia = df[df['Fecha'] == fecha].reset_index(drop=True)

        if df_dia.empty:
            continue

        idx_max = df_dia['S4'].idxmax()
        pico_max = df_dia['S4'].iloc[idx_max]
        umbral = umbral_relativo * pico_max

        # Buscar inicio
        inicio_idx = idx_max
        while inicio_idx > 0 and df_dia['S4'].iloc[inicio_idx] > umbral:
            inicio_idx -= 1

        # Buscar fin
        fin_idx = idx_max
        while fin_idx < len(df_dia) - 1 and df_dia['S4'].iloc[fin_idx] > umbral:
            fin_idx += 1

        inicio_tiempo = df_dia['Tiempo'].iloc[inicio_idx]
        fin_tiempo = df_dia['Tiempo'].iloc[fin_idx]
        duracion_evento = (fin_tiempo - inicio_tiempo).total_seconds() / 60.0  # minutos

        resultados.append({
            'fecha': fecha,
            'inicio_evento': inicio_tiempo,
            'fin_evento': fin_tiempo,
            'pico_max': pico_max,
            'duracion_minutos': duracion_evento
        })
    return resultados
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def detectarMultiplesEventos(df_input, inicio, fin, titulo, umbral_inicio_fin=0.3, umbral_pico=0.6):
    df_input['Tiempo'] = pd.to_datetime(df_input['Tiempo'])
    df = df_input[(df_input['Tiempo'] >= inicio) & (df_input['Tiempo'] <= fin)].reset_index(drop=True)

    eventos = []
    usados = np.zeros(len(df), dtype=bool)

    # Buscar picos locales que superen el umbral de pico
    for idx in range(1, len(df) - 1):
        if (
            df['S4'].iloc[idx] > umbral_pico and
            df['S4'].iloc[idx] > df['S4'].iloc[idx - 1] and
            df['S4'].iloc[idx] > df['S4'].iloc[idx + 1]
        ):
            if usados[idx]:
                continue  # Este pico ya fue cubierto por un evento anterior

            # Buscar inicio del evento (hacia atrás)
            inicio_idx = idx
            while inicio_idx > 0 and df['S4'].iloc[inicio_idx] > umbral_inicio_fin:
                inicio_idx -= 1

            # Buscar fin del evento (hacia adelante)
            fin_idx = idx
            while fin_idx < len(df) - 1 and df['S4'].iloc[fin_idx] > umbral_inicio_fin:
                fin_idx += 1

            # Evitar eventos traslapados
            if np.any(usados[inicio_idx:fin_idx]):
                continue

            # Identificar el pico máximo dentro del rango del evento
            sub_df = df.iloc[inicio_idx:fin_idx]
            pico_idx_local = sub_df['S4'].idxmax()

            eventos.append({
                'inicio': df['Tiempo'].iloc[inicio_idx],
                'fin': df['Tiempo'].iloc[fin_idx],
                'pico': df['S4'].iloc[pico_idx_local],
                'pico_tiempo': df['Tiempo'].iloc[pico_idx_local]
            })

            usados[inicio_idx:fin_idx] = True  # Marcar como usado

    # Graficar todos los eventos
    plt.figure(figsize=(14, 5))
    plt.plot(df['Tiempo'], df['S4'], label='S4', color='blue')

    for i, evento in enumerate(eventos):
        plt.axvline(evento['pico_tiempo'], color='red', linestyle='--', label='Pico máximo' if i == 0 else "")
        plt.axvline(evento['inicio'], color='green', linestyle='--', linewidth=1.2, label='Inicio' if i == 0 else "")
        plt.axvline(evento['fin'], color='orange', linestyle='--', linewidth=1.2, label='Fin' if i == 0 else "")
        plt.fill_between(df['Tiempo'], df['S4'], where=(df['Tiempo'] >= evento['inicio']) & (df['Tiempo'] <= evento['fin']),
                         color='red', alpha=0.2)

    plt.legend()
    plt.title(f"Detección de eventos de centelleo - {titulo}")
    plt.xlabel("Tiempo")
    plt.ylabel("S4")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # Mostrar eventos
    for i, evento in enumerate(eventos):
        print(f"\nEvento {i+1}:")
        print(f" - Inicio : {evento['inicio']}")
        print(f" - Fin    : {evento['fin']}")
        print(f" - Pico S4: {evento['pico']} en {evento['pico_tiempo']}")

    return eventos
```
