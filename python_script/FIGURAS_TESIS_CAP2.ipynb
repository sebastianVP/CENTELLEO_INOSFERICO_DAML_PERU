```python
import numpy as np
import matplotlib.pyplot as plt

# Latitud magnética
lat = np.linspace(-30, 30, 500)

# Modelo simplificado de EIA (dos gaussianas)
crest1 = np.exp(-((lat - 15)/6)**2)
crest2 = np.exp(-((lat + 15)/6)**2)
equator_dip = 0.5 * np.exp(-(lat/8)**2)

Ne = crest1 + crest2 - equator_dip

plt.figure()
plt.plot(lat, Ne)
plt.xlabel("Latitud Magnética (grados)")
plt.ylabel("Densidad Electrónica (normalizada)")
plt.title("Anomalía de Ionización Ecuatorial (EIA)")
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid espacio-temporal
x = np.linspace(0, 1000, 300)   # tiempo
y = np.linspace(200, 600, 300)  # altitud
X, Y = np.meshgrid(x, y)

# Fondo (plasma estable)
background = np.exp(-(Y-300)/200)

# Burbujas ascendentes (EPBs)
bubble1 = np.exp(-((X-300)/80)**2 - ((Y-350)/60)**2)
bubble2 = np.exp(-((X-600)/100)**2 - ((Y-450)/80)**2)

Ne = background - 0.7*bubble1 - 0.7*bubble2

plt.figure()
plt.imshow(Ne, extent=[0,1000,200,600], aspect='auto')
plt.xlabel("Tiempo")
plt.ylabel("Altitud (km)")
plt.title("Burbujas de Plasma Ecuatorial (EPBs)")
plt.colorbar(label="Densidad Electrónica")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Tiempo
t = np.linspace(0, 10, 1000)

# Señal base
signal = 1.0

# Ruido + centelleo
scintillation = 0.3*np.sin(20*t) + 0.2*np.sin(50*t)
noise = 0.1*np.random.randn(len(t))

received_signal = signal + scintillation + noise

plt.figure()
plt.plot(t, received_signal)
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud Normalizada")
plt.title("Centelleo Ionosférico en Señal GNSS")
plt.grid()

plt.show()
```
